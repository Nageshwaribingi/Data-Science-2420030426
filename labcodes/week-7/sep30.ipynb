```python
import pandas as pd 
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

# load iris dataset
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target

#1 Univariate Analysis
#histogram
plt.figure(figsize=(6,4))
sns.histplot(data=df, x='sepal length (cm)', bins=20, kde=True)
plt.title('histogram of Sepal Length')
plt.show()
```

```python
#boxplot 
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x=df['sepal length (cm)'])
plt.title('Boxplot of Sepal Length')
plt.show()
```

```python
#pie chart (species distribution )
species_counts = df['species'].value_counts()
plt.figure(figsize=(6,6))
plt.pie(species_counts, labels=iris.target_names, autopct='%1.1f%%')
plt.title('Species Distribution (pie chart)')
plt.show()
```

```python
# Bivariate visualization
#scatter plot
plt.figure(figsize=(6,4))
sns.scatterplot(x=df['sepal length (cm)'], y=df['sepal width (cm)'], hue=df['species'])
plt.title('Scatter Plot of Sepal Length vs Sepal Width')
plt.show()
```

```python
#line chart 
plt.figure(figsize=(6,4))
plt.plot(df['sepal length (cm)'])
plt.title('Line Plot of Sepal Length')
plt.xlabel('Sample Index')
plt.ylabel('Sepal Length (cm)')
plt.show()
```

```python
#bar graph (mean sepal length by species)
plt.figure(figsize=(6,4))
sns.barplot(x=df['species'], y=df['sepal length (cm)'])
plt.title('Bar chart : Mean Sepal Length by Species')
plt.xticks(ticks=[0, 1, 2], labels=iris.target_names)
plt.show()
```

```python
#multivariate visualization
#heatmap of correlation matrix
plt.figure(figsize=(8,6))
sns.heatmap(df.iloc[:, :4].corr(), annot=True, cmap='coolwarm')
plt.title('Heatmap of Feature Correlation')
plt.show()
```

```python
#bubble chart (scatter plot with size = petal length)
plt.figure(figsize=(6,4))
sns.scatterplot(x=df['sepal length (cm)'], y=df['sepal width (cm)'], s=df['petal length (cm)']*20 , alpha=0.5, c=df['species'])
plt.title('Bubble Chart: Sepal Length vs Sepal Width (Size = Petal Length)')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.show()
```

```python
#pair plot
sns.pairplot(df.iloc[:,:4])
plt.suptitle("Pair Plot of Iris Features",y=1.02)
plt.show()
```

```python
#advanced techniques
#violin plot 
plt.figure(figsize=(6,4))
sns.violinplot(x=df['species'], y=df['sepal length (cm)'])
plt.title('Violin Plot: Sepal Length by Species')
plt.xticks(ticks=[0, 1, 2], labels=iris.target_names)
plt.show()
```
